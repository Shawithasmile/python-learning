การเรียงลำดับตาม keys หรือตามค่า value

```python
#การเรียงลำดับตาม keys หรือตามค่า values
help(sorted)
a={'b': 12, 'a': 18, 'f': 14, 'd': 12, 'e': 18, 'c': 10}
print('--------------------')
print("Ordered by Keys")
for k in sorted(a.keys()):
    print(k,a[k])
print('--------------------')
print("Ordering Values")
for i in sorted(a.values()):
    print(i)
print('--------------------')
print("Ordered by Values")
for i,j in sorted(a.items(),key=lambda x:x[1]):
    print(i,j)
```

```python
student_info = {}

print("กรุณาป้อนข้อมูลในรูปแบบ 'ชื่อ คะแนน' (ป้อน 000 เพื่อหยุด):")
while True:
    data = input()
    if data == "000":
        break
    parts = data.split()
    if len(parts) == 2:
        name = parts[0]
        score = int(parts[1])
        student_info[name] = score
    else:
        print("รูปแบบไม่ถูกต้อง กรุณาป้อนใหม่")

print("\n--- ผลลัพธ์ ---")

print("\n1. เรียงลำดับตามตัวอักษร (a-z):")
for name, score in sorted(student_info.items()):
    print(f"{name} {score}")

print("\n2. เรียงลำดับตามคะแนน (น้อยไปมาก):")
for name, score in sorted(student_info.items(), key=lambda item: item[1]):
    print(f"{name} {score}")

print("\n3. ข้อมูลคะแนนสูงสุดและคะแนนเฉลี่ย:")
if student_info:
    top_student = max(student_info, key=student_info.get)
    average_score = sum(student_info.values()) / len(student_info)

    print(f"นักเรียนที่ได้คะแนนสูงสุด: {top_student} ({student_info[top_student]} คะแนน)")
    print(f"คะแนนเฉลี่ยของทั้งชั้น: {average_score:.2f}")
else:
    print("ไม่มีข้อมูลนักเรียน")
```

ให้แสดงผลของค่า union intersection difference

```python
set1 = set(map(int, input("ป้อนสมาชิกเซตที่ 1 (เว้นวรรคด้วยช่องว่าง): ").split()))
set2 = set(map(int, input("ป้อนสมาชิกเซตที่ 2 (เว้นวรรคด้วยช่องว่าง): ").split()))

union_set = set1 | set2
intersection_set = set1 & set2
difference_set1_set2 = set1 - set2
difference_set2_set1 = set2 - set1

print("\n--- ผลลัพธ์ ---")
print(f"Union (รวมสมาชิกทั้งหมด): {union_set}")
print(f"Intersection (สมาชิกที่ซ้ำกัน): {intersection_set}")
print(f"Difference (เซตที่ 1 - เซตที่ 2): {difference_set1_set2}")
print(f"Difference (เซตที่ 2 - เซตที่ 1): {difference_set2_set1}")
```

__Q3__ ให้สร้าง Dictionary เพื่อนับจำนวนของสมาชิกแต่ละตัวในลิสต์ แล้วแสดงว่า มีสมาชิกอะไรบ้าง และสมาชิกแต่ละตัวมีจำนวนเท่าใด<br/>
a=['1',3,'a',5,6,'a','2',2,'2',2,14,'m','a','b','k','z',7,13, 3,'1','2']<br/>
โดย key ของ Dictionary จะเป็นไอเทมแต่ละตัวในลิสต์ และ values เป็นจำนวนที่ปรากฎในลิสต์

```python
a = ['1', 3, 'a', 5, 6, 'a', '2', 2, '2', 2, 14, 'm', 'a', 'b', 'k', 'z', 7, 13, 3, '1', '2']

count_dict = {}

for item in a:
    if item in count_dict:
        count_dict[item] += 1
    else:
        count_dict[item] = 1

print("จำนวนสมาชิกแต่ละตัวในลิสต์:")
for key, value in count_dict.items():
    print(f"{repr(key)} มีจำนวนสมาชิก {value} ตัว")
```
