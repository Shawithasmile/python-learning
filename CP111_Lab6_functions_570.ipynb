## EX4 Reverse String

Write a Python function `reverse_string` that takes a string as input and returns the reversed string.

```python
def reserve_string(input_string):
  return input_string[::-1]

input_str = "hello"
reversed_str = reserve_string(input_str)
print(reversed_str)
```

## EX5 Palindrome

Create a Python function `is_palindrome` that checks if a given string is a palindrome (Ex. นอน ยาย กนก madam noon kayak level).

```python
# Write your function here
def is_palindrome(string):
  return string == string[::-1]
```

```python
# Example usage:
result = is_palindrome("racecar")
print(result)

print(is_palindrome("Hola"))
```

## EX6 Find Max
Write a Python function `find_max` that finds the maximum element in a list.



```python
# Write your function here
def find_max(numbers):
  max_num = numbers[0]
  for num in numbers:
    if num > max_num:
      max_num = num
  return max_num
```

```python
# Example usage:
numbers = [5, 2, 9, 1, 7]
result = find_max(numbers)
print(result)
```

## Question11
Create a function `sum_numbers` that takes an **arbitrary number of numbers** as arguments and returns their sum.

```python
# Write your function here
def sum_numbers(*numbers):
  sum = 0
  for i in numbers:
    sum += i
  return sum
```

```python
# Example usage:
result = sum_numbers(1, 2, 3, 4, 5)
print(result)  # Output: 15
```

## EX12
Write a function `calculate_total` that takes an **arbitrary number of items and their prices** as keyword arguments.

```python
# Write your function here
def calculate_total(**items):
  print("Item and Prices:")
  total = 0
  for key, price in items.items():
    print(f"- {key}: ${price:.2f}")
    total += price
  print("----------------")
  print(f"Total Price: ${total:.2f}")
  return total
```

```python
# Example usage:
total = calculate_total(apple=1.00, banana=0.50, orange=0.75, milk=2.50)
```
