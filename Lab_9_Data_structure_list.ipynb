Q1 ให้เขียนคำสั่ง เพื่อเติมค่าลงในลิสต์ โดยมีการทำงานดังนี้
<ol>
  <li>เขียนฟังชัน special_insert(list_name, item, No, pos) ทำหน้าที่แทรกค่าลงในลิสต์ โดยมีอากิวเมนต์ดังนี้
      <ul>
          <li>list_name: ชื่อลิสต์</li>
          <li>item: ค่าทีต้องการแทรกลงในลิสต์</li>
          <li>No: จำนวนที่ต้องการแทรก</li>
          <li>pos: ตำแหน่งที่ต้องการแทรก</li>
      </ul>
      </li>
  <li>ตย.
      <ul>
          <li>x=[1,2,3,4]</li>
          <li>a=special_insert(x,[1,2,3],2,2)</li>
          <li>จะได้ผลเป็น a = [1,2,[1,2,3],[1,2,3],3,4]</li>          
      </ul>
  </li>
  <li>ให้ทำสองวิธี วิธีแรกใช้แค่ list concatenation วิธีที่สอง ให้เลือกใช้เฉพาะเมทธอด insert, append หรือ extend</li>
</ol>


```python
def special_insert(list_name, item, No, pos):
    front = list_name[:pos]
    inserted = [item] * No
    back = list_name[pos:]
    return front + inserted + back

x = [1, 2, 3, 4]
a = special_insert(x, [1, 2, 3], 2, 2)
print("วิธีที่ 1:",a)
```

```python
def special_insert(list_name, item, No, pos):
    result = list_name.copy()
    for _ in range(No):
        result.insert(pos, item)
    return result
x = [1, 2, 3, 4]
a = special_insert(x, [1, 2, 3], 2, 2)
print("วิธีที่ 2:", a)
```

```python
def special_insert(list_name, item, No, pos):
    front = list_name[:pos]
    inserted = item * No
    back = list_name[pos:]
    return front + inserted + back
x = [1, 2, 3, 4]
a = special_insert(x, [1, 2, 3], 2, 2)
print(a)
```

len(matrix_a)=? if matrix_a=[[4.0, 6.0, 8.0, 10.0], [4.0, 6.0, 8.0, 10.0]], then we have No. of rows = 2, No. of columns = 4<br/>
How can we determine the number of rows and columns from matrix_a?(let's use len())<br/>
Try using list comprehension so that the code could be cleaner



```python
สำหรับเมทริกซ์ในรูปแบบ Nested List เช่น matrix_a = [[4.0, 6.0, 8.0, 10.0], [4.0, 6.0, 8.0, 10.0]]:
หาจำนวนแถว (Number of Rows):
 rows = len(matrix_a)
 หาจำนวนคอลัมน์ (Number of Columns):
 cols = len(matrix_a[0])
```

```python
def input_matrix(name):
    print(f"\n--- ป้อนข้อมูลสำหรับเมทริกซ์ {name} ---")
    rows = int(input(f"กรุณาป้อนจำนวนแถวของเมทริกซ์ {name}: "))
    matrix = []
    for i in range(rows):
        row_input = input(f"ป้อนข้อมูลแถวที่ {i + 1} (เว้นช่องว่างระหว่างตัวเลข): ")
        row = [int(x) for x in row_input.split()]
        matrix.append(row)
    return matrix

matrix_A = input_matrix("A")
matrix_B = input_matrix("B")
rows_A = len(matrix_A)
cols_A = len(matrix_A[0]) if rows_A > 0 else 0

rows_B = len(matrix_B)
cols_B = len(matrix_B[0]) if rows_B > 0 else 0

print("\n--- ผลการประมวลผล ---")
if rows_A != rows_B or cols_A != cols_B:
    print("ไม่สามารถบวกกันได้ เนื่องจากมิติของเมทริกซ์ไม่เท่ากัน")
    print(f"มิติของ A: {rows_A}x{cols_A}, มิติของ B: {rows_B}x{cols_B}")
else:
    result = []
    for i in range(rows_A):
        row_result = []
        for j in range(cols_A):
            sum_val = matrix_A[i][j] + matrix_B[i][j]
            row_result.append(sum_val)
        result.append(row_result)

    print("ผลลัพธ์การบวกเมทริกซ์ (A + B):")
    print(result)
```
