### Structure of Try and Exception block

Example of using try and exception to prevent error messages

```python
q = 'N/A'
try:
    a = float(input('enter first number here:'))
    b = float(input('enter second number here:'))
    c = a / b
    q = c
except ZeroDivisionError:
    q = 'N/A'
except ValueError:
    print("คุณกรอกข้อมูลไม่ใช่ตัวเลข")
finally:
    print(q)
```

```python
q = 'N/A'
while True:
    try:
        a = float(input('enter first number here:'))
        b = float(input('enter second number here:'))
        c = a / b
    except ZeroDivisionError:
        q = 'N/A'
        break
    except ValueError:
        print('Input number please')
        continue
    else:
        q = c
        break
    finally:
        print(q)
```

```python
file_content = """John 18 170 45
Abbel 30 190 50
Pat 24 172 48
Lindsey 29 170 49"""
with open("student_info.txt", "w", encoding="utf-8") as f:
    f.write(file_content)
print("สร้างไฟล์ student_info.txt เรียบร้อยแล้ว!")
```

```python
while True:
    file_name = input("input file name: ")
    try:
        with open(file_name, "r", encoding="utf-8") as f:
            lines = f.readlines()
        break
    except FileNotFoundError:
        print("no such file, please re-input file name")
records = []
for line in lines:
    data = line.split()
    if len(data) >= 4:
        name = data[0]
        age = int(data[1])
        height = float(data[2])
        weight = float(data[3])
        bmi = weight / ((height / 100) ** 2)
        records.append({
            "name": name,
            "age": age,
            "height": int(height),
            "weight": int(weight),
            "bmi": bmi
        })
valid_fields = ["name", "age", "height", "weight", "bmi"]
valid_ops = ["=", ">", "<"]
while True:
    try:
        query = input("query: ")
    except EOFError:
        break
    parts = query.strip().split()
    if len(parts) != 3:
        print("Wrong query format")
        continue
    field_raw, op, val_str = parts
    field = field_raw.lower()
    if op not in valid_ops:
        print("Wrong query format")
        continue
    if field not in valid_fields:
        print("No such query")
        continue
    try:
        if field == "name":
            val = val_str
        else:
            val = float(val_str)
    except ValueError:
        print("Wrong query format")
        continue
    found = False
    for rec in records:
        rec_val = rec[field]
        match = False
        if field == "name":
            rec_val_str = rec_val.lower()
            val_str_lower = str(val).lower()
            if op == "=" and rec_val_str == val_str_lower:
                match = True
            elif op == ">" and rec_val_str > val_str_lower:
                match = True
            elif op == "<" and rec_val_str < val_str_lower:
                match = True
        else:
            if op == "=" and rec_val == val:
                match = True
            elif op == ">" and rec_val > val:
                match = True
            elif op == "<" and rec_val < val:
                match = True
        if match:
            print(f"{rec['name']} {rec['age']} {rec['height']} {rec['weight']} {rec['bmi']:.2f}")
            found = True
    if not found:
        print("No such person")
```
