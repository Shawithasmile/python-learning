## Question#6: Filtering Even Nubmers

Use a lambda function to filter even numbers from a list.

```python
numbers = [1, 2, 3, 4, 5, 6]
even_numbers = list(filter(lambda x: x%2 == 0, numbers))
print(even_numbers)  # Output: [2, 4, 6]
```

## Question#7: Mapping Strings to Lengths
Create a lambda function to map a list of strings to their corresponding lengths.



```python
words = ["apple", "banana", "cherry"]
lengths = list(map(lambda word: len(word), words))
print(lengths)  # Output: [5, 6, 6]
```

# Question#8: Sorting a List of Tuples
Use a lambda function to sort a list of tuples by their second element.

```python
tuples = [('ดำเกิง',29), ('บุญหลาย', 22), ('ละอองดาว', 15)]
sorted_tuples = sorted(tuples, key=lambda x: x[1])
print(sorted_tuples)  # Output: [('ละอองดาว', 15), ('บุญหลาย', 22), ('ดำเกิง',29)]
```

#Question#9: Creating a Power of Two Function with a Lambda Expression
Write a function that takes a list and a function as arguments.
To implement this function, calculates the power of two for each argument, and returns the results.

```python
def apply_func(func, items):
    return list(map(func, items))

numbers = [1, 2, 3]
squared_numbers = apply_func(lambda x: x**2, numbers)
print(squared_numbers)  # Output: [1, 4, 9]

```

#Question#10: Power Function
Create a recursive function to calculate the power of a number.

```python
def power(base, exponent):
    if exponent == 0:
      return 1
    elif exponent < 0:
      return 1/power(base, -exponent)
    return base* power(base, exponent -1)
print(power(2, 10))
```
