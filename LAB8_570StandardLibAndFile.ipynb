## ตำแหน่งของ file pointer ตอนเปิดไฟล์ด้วย a+, w+, r+ mode
a+ ตำแหน่งไฟล์พอยเตอร์อยู่ที่จุดสิ้นสุดไฟล์ เมื่อสั่ง f.write() จะปรากฎที่ท้ายไฟล์<br/>
r+ ตำแหน่งไฟล์พอยเตอร์อยู่ที่จุดเริ่มต้นไฟล์ เมื่อสั่ง f.write() จะปรากฎที่ต้นไฟล์<br/>
w+ ตำแหน่งไฟล์พอยเตอร์อยู่ที่จุดเริ่มต้นไฟล์ เมื่อสั่ง f.write() จะปรากฎที่ต้นไฟล์และเขียนทับข้อมูลเดิมทั้งหมด<br/>

```python
f = open('newfile.txt','w')
f.write('this is my first line \n')
f.write('this is my second line \n')
f.write('Is this my third line? \n')
f.close()

f = open('newfile.txt','a+')
f.write('a new line is added')
print(f.read())
#Notice that there will be nothing displayed because file pointer is at the end of the file
f.seek(0,0)
#to read the file we set the file pointer back to the start point by f.seek()
print('\n write in \'a+\' mode:')
print(f.read())
f.close()
```

## กำหนดตำแหน่งของไฟล์พอยเตอร์ด้วย f.seek
f.seek(offset,location)

```python
f = open('newfile.txt','w')
f.write('this is my first line \n')
f.write('this is my second line \n')
f.write('Is this my third line? \n')
f.close()

f = open('newfile.txt','a+')
f.write('a new line is added')
print('when we read without using f.seek()\n')
print(f.read())
#Notice that there will be nothing displayed because file pointer is at the end of the file

f.seek(0,0)
print('when we read using f.seek() first\n')
print(f.read())
#file pointer will go back to the starting point of the file
f.close()
```

Opneing a file in 'r+' mode, when setting the file pointer to the end of file, we can append the new content in the same way as when we open in 'a+' mode.

```python
f = open('newfile.txt','w')
f.write('this is my first line \n')
f.write('this is my second line \n')
f.write('Is this my third line? \n')
f.close()

f = open('newfile.txt','r+')
f.seek(0,2)
f.write('a new line is added')
f.seek(0,0)
print(f.read())
f.close()
```

```python
#Q1.1
#create a file
f = open('newfile.txt', 'w')
for i in range(1, 11):
    f.write(str(i) + '\n')
f.close()

f = open('newfile.txt', 'r')
content = f.read()
f.close()

updated_content = content.replace('7', '6')
f = open('newfile.txt', 'w')
f.write(updated_content)
f.close()

print("=== ผลลัพธ์ข้อมูลล่าสุดในไฟล์ newfile.txt (Q1.2) ===")
f = open('newfile.txt', 'r')
print(f.read())
f.close()
```

```python
#Q1.2
#modify the file using method replace() in string
# Q1.1
with open('newfile.txt', 'w+') as f:
    for i in range(1, 11):
        f.write(f"{i}\n")

    f.seek(0)
    content = f.read()

    updated_content = content.replace('7', '6')

    f.seek(0)
    f.truncate()
    f.write(updated_content)

    f.seek(0)
    print("=== ผลลัพธ์ในไฟล์ newfile.txt หลังแก้ไข (Q1.1) ===")
    print(f.read())
```
