## Question#1 หาค่ารากที่สองของจำนวนใดๆ

## Question#2 คำนวณหาพื้นที่ของวงกลม
เขียนโปรแกรม Python ที่รับค่ารัศมีของวงกลมจากผู้ใช้ แล้วคำนวณหาพื้นที่ของวงกลม โดยใช้ค่า `pi` จากโมดูล `math`



```python
import math
radius = float(input("กรอกรัศมีของวงกลม:"))
area = math.pi * (radius ** 2)
print(f"พื้นที่วงกลมของ {radius} คือ {area:.2f}")
```

# Question#3 เกมทายตัวเลข

สร้างโปรแกรม Python ที่ให้คอมพิวเตอร์สุ่มตัวเลขจำนวนเต็มระหว่าง 1 ถึง 100 แล้วให้ผู้ใช้ทายตัวเลขนั้น หากผู้ใช้ทายผิด ให้บอกว่ามากเกินไปหรือน้อยเกินไป และให้ผู้ใช้ทายใหม่ จนกว่าจะทายถูก

**ตัวอย่าง**

image.png

```python
import random
min_num, max_num =1,100
secret_numbers = random.randint(min_num,max_num)
print(f"คิดเลขไว้แล้วนะ ระหว่าง {min_num} ถึง {max_num}")
print("ลองทายสิ")
while True:
    guess = input("ทายเลข:").strip()
    if not guess.isdigit():
      print("ไม่ใช่ตัวเลขนะ ลองใหม่!")
    else:
      guess = int(guess)

    if guess == secret_numbers:
      print("เยี่ยมมาก! คุณทายถูกแล้ว")
      break
    else:
      print("มากไปนะ ลองใหม่" if guess > secret_numbers else "น้อยไปนะ ทายใหม่")
```

## Question#5

เขียนโค้ดหาสตริงย่อยระหว่าง "`<title>`" และ "`</title>`" โดยไม่ใช้ Regular Expression

**แนวทางการแก้ปัญหา**

1. หาตำแหน่งของแท็ก `<title>`:

  * วนลูปผ่านสตริงทีละตัวอักษร
เมื่อเจอ "`<title>`"
  * ให้บันทึกตำแหน่งไว้
2. หาตำแหน่งของแท็ก `</title>`:

  * ต่อจากตำแหน่งที่เจอ "`<title>`" ให้วนลูปหาตำแหน่งของ `</title>`
ตัดสตริง:

3. นำตำแหน่งที่ได้มาตัดสตริงส่วนที่ต้องการ




```python
#เขียน code ที่นี่
#กำหนด HTML Tag ดังนี้
html_string = """
<html>
<head>
<title>My Website</title>
</head>
<body>
</body>
</html>
"""

start_index = html_string.find("<title>")
end_index = html_string.find("</title>", start_index)
if (start_index < 0 ) or (end_index < 0):
  print("ไม่พบ tag <title> หรือ </title>")
else:
  print("title :", html_string[start_index+len("<title>"):end_index])
```

## Question#6

เขียนโปรแกรมที่รับประโยคหนึ่งประโยค แล้วหาคำที่ยาวที่สุดในประโยคนั้น





```python
sentence = "this is a sample sentence to find to longest word"
words = sentence.split()
longest_word = max(words, key=len)
print("คำที่ยาวที่สุดคือ:", longest_word)
```
